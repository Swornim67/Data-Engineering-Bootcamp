```sql
select * from employees
```

```sql
select emp_id,first_name,salary from employees 
```

```sql
--only those persons who are inactive
select * from employees where status='Inactive'
```

```sql
--salary greater then 80k
select * from employees where salary > 80000
```

```sql
--salary greater then 1 lakhs
select * from employees where salary > 100000
```

```sql
-- active and lives in banglore 
select * from employees where status='Active' and city='Bangalore'
```

```sql
-- working in engineering and marketing department 
select * from employees where department in("Engineering",'Marketing')
```

```sql
-- working in engineering and hr department and salary greater then 90000
select * from employees where department in("Engineering",'HR') and salary > 90000
```

```sql
--joined before 2020
select * from employees where hire_date < '2020-01-01'
```

```sql
--highest salary
select * from employees order by salary desc
```

```sql
--lowest salary 
select * from employees order by salary
```

```sql
--department and name alphabetically
select * from employees order by department,first_name 
```

```sql
--recent hired 
select first_name,hire_date from employees order by hire_date desc 
```

```sql
--only 1st row person who got hired recently
select first_name,hire_date from employees order by hire_date desc limit 1
```

```sql
--first 2 rows hired recently
select first_name,hire_date from employees order by hire_date desc limit 2
```

```sql
-- print only the 2nd person of the row who got hired recently
select first_name,hire_date from employees order by hire_date desc limit 1 offset 1
```

```sql
--2nd highest salary 
select * from employees order by salary desc limit 1 offset 1
```

```sql
--3rd highest salary 
select * from employees order by salary desc limit 1 offset 2
```

```sql
select department from employees group by department
```

```sql
--count the no of employees in each department
select department,count(*) as `Total Emp in department` from employees group by 1 
```

```sql
-- count the no of employees less then 9 in department
select department,count(*) as `Total Emp in Department` from employees group by department having count(*) < 9
```

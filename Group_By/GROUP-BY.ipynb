Group By


```sql
--Get the count of employees per department.
select department,count(*) from employees group by department
```

```sql
--Get the average salary per department.
select department,avg(salary) from employees group by department
```

```sql
--Get the highest salary per city.
select city,max(salary) from employees group by city
```

```sql
--Get the count of employees per status (Active/Inactive).
select status,count(*) from employees group by status
```

```sql
--Get the total salary paid out per department.
select department,sum(salary) from employees group by department
```

```sql
--Get the count of employees per department, per status (group by 2 columns).
select department,status,count(*) as `Total Employees` from employees group by department,status
```

```sql
--Get the average salary per department, but only considering 'Active' employees
select department,avg(salary) from employees where status='Active' group by department
```

Having

```sql
--Get departments where the count of employees is greater than 8.
select department,count(*) from employees group by department having count(*) > 8 
```

```sql
--Get departments where the average salary is above 75000.
select department,avg(salary) from employees group by department having avg(salary) > 75000 
```

```sql
--get cities where the highest salary exceeds 100000.
select city,max(salary) from employees group by city having max(salary) > 100000
```

```sql
--Get departments where the total salary paid is above 700000.
select department,sum(salary) from employees group by department having sum(salary) > 700000
```

```sql
--Get departments (only considering 'Active' employees) where average salary is above 70000
select department,avg(salary) from employees where status='Active' group by department having avg(salary) > 70000
```

```sql
--Get statuses where the count is less than 5.
select status,count(*) from employees group by status having count(*) < 5
```

Select Queries 

```sql
--Select all columns from employees.
select * from employees;
```

```sql
--Select only first_name.
select first_name from employees;
```

```sql
--Select first_name and last_name.
select first_name,last_name from employees;
```

```sql
--Select department, job_title, and salary — in that order.
select department,job_title,salary from employees;
```

```sql
--Select the same three columns from Q4, but reverse the order: salary, job_title, department.
select salary,job_title,department from employees;
```

```sql
--Select every column except imagine you "left out" email 
select emp_id, first_name, last_name, department, job_title, salary, hire_date, status, city
from employees;
```

 Aliasing with AS

```sql
--Select first_name, but display the column as Name.
select first_name as name from employees 
```

```sql
--Select last_name, but display it as Surname.
select last_name as surname from employees
```

```sql
--Select email, but display it as Contact_Email.
select email as contact_email from employees
```

```sql
--Select first_name and last_name, aliasing them as First and Last respectively (two aliases in one query).
select first_name as first, last_name as last from employees
```

```sql
--Select department, but display it with a column name that has a space in it: "Dept Name"
select department as `Dept Name` from employees
```

```sql
--Select salary, aliasing it as Annual_Salary 
select salary as Annual_Salary from employees
```

literals & expressions

```sql
--Select first_name, last_name, plus a literal text column called 'Status_Check' that just says 'Pending' for every row.
select first_name,last_name, 'Pending' as status_check from employees
```

```sql
--Select first_name and salary, plus a new column bonus_10pct showing salary * 0.10.
select first_name,salary, (salary *0.10) as bonus_10pct from employees
```

```sql
--Select first_name and salary, plus a new column new_salary showing what salary would be after a 10% raise (salary + salary*0.10, or equivalently salary * 1.10).
select first_name,salary, salary *1.10 as `New_salary After 10% raise` from employees
```

```sql
--Select first_name, salary, and a column monthly_salary showing salary / 12.
select first_name,salary, salary/12 as `Monthly salary` from employees
```

```sql
--Select emp_id, plus a literal number column called bonus_flag that's just the number 1 for every row
select emp_id,1 as bonus_flag from employees
```


String concatenation

```sql
--Select first_name and last_name combined into one column called full_name, with a space between them — use ||.
select first_name ||' '|| last_name as full_name from employees
```

```sql
--Do the same thing as Q1, but using CONCAT() instead.
select concat(first_name,' ',last_name) as full_name from employees 
```

```sql
--Select a column that combines first_name, last_name, and department, separated by a comma and space — like "John Doe, Engineering" — aliased as summary.
select concat('"', first_name, ' ', last_name, ', ', department, '"') as summary from employees
```

```sql
--Select a column combining city and department with a dash between them — like "Bangalore-Engineering" — aliased as location_dept.

select concat('"',city,'-',department,'"') as location_dept from employees
```

Distinct

```sql
--Select the distinct department values from employees.
select distinct department from employees
```

```sql
--Select the distinct city values from employees.
select distinct city from employees
```

```sql
--Select the distinct status values — there should only be 2 (Active/Inactive).
select distinct status from employees
```

```sql
--Select distinct combinations of department and city together.
select distinct department,city from employees
```

```sql
--Select distinct job_title values — how many different job titles are there? (count them once you see the result)
select count(distinct job_title) as different_title from employees
```

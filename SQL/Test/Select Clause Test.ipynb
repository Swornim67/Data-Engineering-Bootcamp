```sql
--Select first_name and last_name concatenated into a column called Full Name (spaced alias), plus a literal column source saying 'internal'.
select concat(first_name,' ',last_name) as Full_name, 'internal' as source from employees 
```

```sql
--Select emp_id, salary, a column New salary (salary after a 15% raise), and a column Raise amount (just the raise itself).
select emp_id,salary, salary+salary * 0.15 as `New Salary`,salary * 0.15 as `Raised Amount` from employees 
```

```sql
--Select the distinct combination of department and status.
select distinct department,status from employees
```

```sql
--Select a column combining job_title, " - ", and city, aliased as posting_label.
select concat(job_title,' - ',city) as posting_label from employees 
```

```sql
--Select first_name, salary, a column Monthly Pay (salary / 12), and a literal column tax_year with value 2024.
select first_name,salary,salary/12 as `Monthly Pay`, 2024 as tax_year from employees
```

```sql
--Select distinct job_title values.
select distinct job_title from employees
```

```sql
--Select first_name, last_name, a concatenated "Last, First" style column aliased as display_name, plus a literal column active_flag with value 1.
select concat(last_name,', ',first_name) as display_name, 1 as active_flag from employees 
```

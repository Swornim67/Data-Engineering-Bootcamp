```sql
--Select first_name, last_name, and a column showing salary after a 10% raise (aliased new_salary), but only for employees in 'Engineering'.
select first_name,last_name,salary,salary+salary * 0.10 as `New Salary` from employees where department='Engineering'
```

```sql
--Select a concatenated full_name column (first + last name) for employees whose salary is between 70000 and 100000.
select concat(first_name,' ',last_name) as full_name,salary from employees where salary between 70000 and 100000
```

```sql
--Select first_name, department, city, but only for employees in 'Hyderabad' or 'Pune'
select first_name,department,city from employees where city in ('Hyderabad' ,'Pune')
```

```sql
--Select first_name, last_name, and a literal column 'Review_Due' for all employees who are 'Active' and earn more than 85000.
select first_name,last_name,'Review_Due ' as review_Due from employees where status='Active' and salary > 85000
```

```sql
--Select the distinct department values, but only considering employees hired after '2020-01-01'
select distinct department from employees where hire_date > '2020-01-01'
```

```sql
--Select first_name, email, for employees whose email contains 'kumar' or 'singh'.
select first_name,email from employees where email like '%kumar%' or email like '%singh%'
```

```sql
--Select first_name, last_name, job_title, for employees whose job_title starts with 'Data'.
select first_name,last_name,job_title from employees where job_title like 'Data%'
```

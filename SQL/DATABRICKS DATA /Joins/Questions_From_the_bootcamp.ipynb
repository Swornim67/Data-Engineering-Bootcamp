```sql
%sql
-- WRITE A SQL QUERY TO GET THE TOTAL SALES PER REGION 
select 
region,sum(amount) as `TOTAL SALES`
from sales 
group by region;
```

```sql
%python
-- RETRIEVE ALL ORDERED PLACED ON OR AFTER 2023-10-02
select product,sale_date,count(*) as `NO OF PRODUCTS ORDERED`
from sales 
where sale_date >='2023-10-02'
group by product,sale_date;
```

```sql
%sql
-- FIND ALL CUSTOMERS WHO MADE PURCHASE GREATER THEN 2000.
SELECT customer_id, amount, product
FROM sales
WHERE amount > 2000;
```

```sql
%sql
-- WRITE A QUERY TO DISPLAY THE HIGHEST SINGLE SALE AMOUNT AND ITS REGION.
SELECT amount AS `HIGHEST SALE AMOUNT`, region
FROM sales
ORDER BY amount DESC
LIMIT 1;

```

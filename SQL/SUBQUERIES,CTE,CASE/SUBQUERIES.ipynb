SUBQUERIES

```sql
-- FIND CUSTOMERS WHO PLACED ORDERS ABOVE THE AVERAGE ORDER AMOUNT.

select distinct c.first_name,c.last_name,o.total_amount
from customers c 
join orders o 
on c.customer_id=o.customer_id
where o.total_amount >
(
select avg(total_amount) from orders
)
```

```sql
-- LIST ALL PRODUCTS THAT HAVE BEEN SOLD MORE THAN 1 TIMES IN TOTAL.
select 
product_name,total_sold
from

(
select 
product_name,sum(quantity) as total_sold
from order_details 
group by product_name)

where total_sold > 1;
```

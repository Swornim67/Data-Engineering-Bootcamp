```python
-- Question 1
-- Find all orders where the total_amount is greater than the average order amount across all orders.

SELECT *
from customers c 
join orders o 
on c.customer_id=o.customer_id
where o.total_amount >

(
select avg(total_amount)from orders

)
```

```python
-- Question 2
-- Find the customer(s) who placed the single highest-value order.

select *
from customers c 
join orders o
on c.customer_id=o.customer_id
where o.total_amount =
(
    select max(total_amount) from orders
);
```

```python
-- Question 3
-- Find all customers whose city is the same as customer_id 1's city .

select first_name,last_name,city 
from customers 
where city =
(
    select city from customers where customer_id=1
);
```

```python
-- Question 4
-- Find all orders placed after the date of order_id 1005 (2023-02-05).

select * from orders 
where order_date > 
(
    select order_date from orders 
    where order_id=1005
);
```

```python
-- Question 5
-- Find all customers who have placed at least one order

select distinct * from customers 
where customer_id in 
(
    select customer_id from orders
);
```

```python
-- Question 6
-- Find all customers who have never placed an order (using NOT IN — watch out for the NULL trap).

select *
from customers 
where customer_id not in 
(
    select customer_id from orders
)
```

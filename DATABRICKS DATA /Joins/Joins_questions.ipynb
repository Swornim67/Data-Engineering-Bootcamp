INNER JOIN QUESTIONS.

```sql
-- list all the customers orders with product details.
select 
c.customer_id,c.first_name,c.last_name,o.order_id,o.product_name,o.quantity
from customers c 
join order_details o 
on c.customer_id=o.customer_id;
```

```sql
%sql
-- show me full breakdown : customers,orders,product_info 
select 
c.customer_id,c.first_name,od.product_name,od.quantity,od.price_each
from order_details od
join orders o on od.order_id=o.order_id
join customers c  on o.customer_id=c.customer_id
```

LEFT JOIN

```sql
-- show inactive and active customers (people who are ordering considered as active customers and who doesnt order are considered as inactive customers).
select 
c.customer_id,c.first_name,c.last_name,
case 
when o.order_detail_id is null then 'Inactive'
else'Active'
end as customer_status
from customers c 
left join order_details o
on c.customer_id=o.customer_id
```

```sql
-- customer who never placed an order(placed those person in the column named inactive customers also if product_id is null then make it as no product_id)
select 
c.customer_id,c.first_name,c.last_name,
COALESCE(cast(o.product_id as string),'No Product ID') as product_id,
case 
when o.product_id is null then 'Inactive'
else 'Active'
end as customer_status
from customers c
left join order_details o
on c.customer_id=o.customer_id
```

RIGHT JOIN


```sql
-- Show me the inactive and active customers(inactive is someone who didnt ordered anything and active is someone who have ordered things.)
select 
c.customer_id,c.first_name,c.last_name,
coalesce(cast(o.order_date as string),'No order date') as `Order Date`,
case 
when o.order_detail_id is null then 'Inactive'
else 'Active'
end as customer_status
from order_details o
right join customers c
on o.customer_id=c.customer_id


```

```sql
-- For the Self Join Making A new table.
CREATE TABLE emp (
    emp_id INT ,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    manager_id INT,
    department VARCHAR(50)
);
```

```sql
-- FIND THE LIST OF PERSON HAVING MANAGERS AND NO MANAGERS.
select 
e.first_name,
coalesce(cast(m.first_name as string),'No Manager') as `Manager Name`,e.department
from emp e
left join  emp m
on e.manager_id=m.emp_id
```

```sql
-- FIND DIRECT REPORTING MANAGERS COUNTS.
select
m.first_name as `Manager Name`,count(e.emp_id) as `No of Reporting to Managers`
from emp e
join emp m
on e.manager_id=m.emp_id
group by m.first_name
order by count(e.first_name) desc;
```

FULL OUTER JOIN


```sql
-- FULL AUDIT OF CUSTOMERS ORDERS + ORDERS(MATCHED OR NOT)
select 
c.first_name,c.last_name,o.order_id,od.product_name
from customers c 
full outer join orders o on c.customer_id=o.customer_id
full outer join order_details  od on o.order_id=od.order_id
```

```sql
-- GET LIST OF ALL THE UNIQUE CUSTOMERS CITIES + UNIQUE PRODUCT CATEGORIES.
select 
distinct city as item from customers
union 
select distinct category from order_details
```

```sql
-- GET LIST OF ALL THE  CUSTOMERS CITIES + PRODUCT CATEGORIES.
select distinct city from customers 
union all
select distinct category from order_details
```

```sql
-- top 3 customers by spent ( we can use either order table or order details table.)
select 
c.first_name,c.last_name,sum(od.quantity*od.price_each) as `TOTAL SPENT`
from customers c 
join order_details od 
on c.customer_id=od.customer_id
group by c.first_name,c.last_name
order by `TOTAL SPENT` desc limit 3;
```

ORDER BY


```sql
--Get all employees sorted by salary, ascending (lowest first).
select first_name,last_name,salary from employees order by salary
```

```sql
--Get all employees sorted by salary, descending (highest first).
select first_name,last_name,salary from employees order by salary desc
```

```sql
--Get all employees sorted alphabetically by first_name.
select first_name from employees order by first_name
```

```sql
--Get all employees sorted by hire_date, most recent first.
select first_name,last_name,hire_date from employees order by hire_date desc
```

```sql
--Get all employees in 'Engineering', sorted by salary descending.
select first_name,last_name,department,salary from employees where department='Engineering' order by salary desc
```

```sql
--Sort all employees by department ascending, and within each department, by salary descending.
select first_name,last_name,department,salary from employees  order by  department asc,salary desc
```

Finishing ORDER BY

```sql
--Get first_name, department, salary — sort by the 3rd column descending
select first_name,department,salary from employees order by 3 desc
```

```sql
--Get first_name, salary, and a column bonus (salary * 0.10) — sort by the bonus alias, descending.
select first_name,salary,salary*0.10 as bonus from employees order by bonus desc
```

```sql
--Get first_name, city — sort by city, explicitly saying NULLS LAST 
select first_name,city from employees order by city Nulls last
```

```sql
--Get first_name, department, hire_date — sort by department using position number, ascending.
select first_name,department,hire_date from employees order by 2 asc
```

```sql
--Get the total number of employees.
select count(*) as `Total Employees` from employees 
```

```sql
--Get the average salary across all employees.
select avg(salary) as `Avg_salary` from employees
```

```sql
--Get the highest salary in the company.
select max(salary) as `Max salary` from employees
```

```sql
--Get the lowest salary in the company.
select min(salary) as `Min_salary` from employees
```

```sql
--Get the total (sum) of all salaries combined.
select sum(salary) as `total salary ` from employees
```

```sql
--Get the average salary of employees in 'Engineering' only.
select avg(salary) from employees where department='Engineering'
```

```sql
--Get the count of employees who are 'Active'.
select count(*) from employees where status='Active'
```

```sql
--Get MIN, MAX, and AVG salary all in one query.
select min(salary),max(salary),avg(salary) from employees
```

Final Stress Test — SELECT + WHERE + GROUP BY + HAVING + ORDER BY + LIMIT

```sql
--Get each department along with the count of employees in it, but only for employees whose status = 'Active', only showing departments with more than 7 active employees, sorted by count descending.

select department,count(*) from employees where status='Active' group by department having count(*) > 7  order by count(*) desc
```

```sql
--Get each city with its average salary, rounded conceptually (just use plain AVG for now), but only include cities where the average salary is above 70000, and only consider employees hired after '2018-01-01' — sorted by average salary descending, top 3 only.

select city, avg(salary)from employees where hire_date > '2018-01-01' group by city having avg(salary) > 70000 order by avg(salary) desc limit 3
```

```sql
--Get each department's highest paid employee's salary (MAX(salary)), but exclude the 'HR' department entirely, only show departments where that max salary exceeds 95000, sorted alphabetically by department.

select department,max(salary)from employees where department !='HR' group by department having max(salary) > 95000 order by department asc
```

```sql
--Get status along with total salary paid per status, but only for employees in ('Engineering', 'Finance', 'Sales'), sorted by total salary descending.

select status,sum(salary) from employees where department in('Engineering','Finance','Sales') group by status order by sum(salary) desc
```

```sql
--Get each department with its employee count, but only include departments where the count is between 5 and 10
select department,count(*) from employees group by department having count(*) between 5 and 10 
```

```sql
--Get each city, count of employees, and average salary — all three in one query — only for 'Active' employees, only cities with more than 4 employees, sorted by average salary descending, limited to top 5.

select city,count(*) as `Emp Count`,avg(salary) from employees where status='Active' group by city having count(*) > 4 order by avg(salary) desc limit 5 
```

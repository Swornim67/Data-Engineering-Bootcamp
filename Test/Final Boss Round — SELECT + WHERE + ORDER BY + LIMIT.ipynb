```sql
--Get the first_name, last_name, and a column full_name (concatenated, "First Last" format) for employees whose salary is above 70000 and whose department is either 'Engineering' or 'Finance' — sorted by salary descending, showing only the top 5.

select first_name,last_name,concat(First_name,' ',last_name) as Full_name,salary,department from employees where salary > 70000 and department in('Engineering','Finance') order by salary desc limit 5
```

```sql
--Get first_name, department, city, salary for employees who are NOT in ('Sales', 'Marketing'), whose first_name does not start with 'S', sorted alphabetically by first_name.

select first_name,department,city,salary from employees where department not in('Sales','Marketing') and first_name not like 'S%' order by first_name 
```

```sql
--Get first_name, last_name, hire_date for employees hired between '2018-01-01' and '2021-12-31', whose job_title contains the word 'Engineer' anywhere, sorted by hire_date ascending — show only the first 4 results.

select first_name,last_name,hire_date from employees where hire_date between '2018-01-01' and '2021-12-31' and job_title like '%Engineer%' order by hire_date asc limit 4 
```

```sql
--Get first_name, department, salary, and a computed column bonus (10% of salary) for employees where (department = 'Engineering' and salary > 85000) or (department = 'Finance' and salary > 65000) — sorted by bonus descending.

select first_name,department,salary,salary*0.10 as bonus from employees where (department = 'Engineering' and salary > 85000) or (department = 'Finance' and salary > 65000) order by bonus desc 
```

```sql
--Get emp_id, first_name, city for employees whose city is IN ('Bangalore', 'Chennai', 'Hyderabad') and whose email does not contain 'kumar' and whose status is 'Active' — sorted by city ascending, then first_name ascending, limited to 8 rows.

select emp_id,first_name,city,email,status from employees where city in('Bangalore', 'Chennai', 'Hyderabad') and email not like '%kumar%' and status='Active' order by city asc,first_name asc limit 8 
```

```sql
--Get first_name, last_name, salary, and a literal column pay_grade that just says 'Senior' for every row — but only for employees whose salary is between 90000 and 130000 and whose hire_date is before '2019-01-01' — sorted by salary descending, top 3 only.

select first_name,last_name,salary,'Senior' as pay_grade from employees where salary between 90000 and 130000 and hire_date <'2019-01-01' order by salary desc limit 3 
```

```sql
--get first_name, department, salary for employees in 'Engineering' sorted by salary descending, but then get the row at position 3 in that sorted list only

select first_name,department,salary from employees where department='Engineering' order by 3 desc limit 1 offset 2
```

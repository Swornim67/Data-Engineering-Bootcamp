Basic Where Clause


```sql
--Get all employees in 'Marketing'.
select first_name,last_name,department from employees 
where department= 'Marketing'
```

```sql
--Get all employees with salary greater than 80000.
select first_name,last_name,salary from employees 
where salary > 80000;
```

```sql
--Get all employees with salary less than or equal to 60000.
select first_name,last_name,salary from employees 
where salary <=60000
```

```sql
--Get all employees who are not in 'Engineering' (use != or <>).
select first_name,last_name,department from employees 
where department !='Engineering'
```

```sql
--Get all employees hired before '2019-01-01'.
select * from employees where hire_date < '2019-01-01'
```

```sql
--Get all employees with emp_id equal to 25.
select * from employees where emp_id=25
```

AND,OR,NOT QUESTIONS


```sql
--Get employees in 'Engineering' AND with salary > 90000.
SELECT first_name,last_name,department,salary FROM EMPLOYEES WHERE DEPARTMENT='Engineering' and salary > 90000
```

```sql
--Get employees in 'HR' OR 'Finance'.
select first_name,last_name,department from employees where department = 'HR' or department = 'Finance'
```

```sql
--Get employees who are NOT 'Active'.
select first_name,last_name,status from employees where status !='Active'
```

```sql
--Get employees in 'Sales' AND hired after '2020-01-01'.
select first_name,last_name,department,hire_date from employees where department='Sales' and hire_date >'2020-01-01'
```

```sql
--Get employees who are in 'Engineering' OR earn more than 100000 
select First_name,last_name,department,salary from employees where department ='Engineering' or salary > 100000
```

```sql
--Get employees in 'Marketing' AND city is 'Delhi' AND salary above 55000 — three conditions combined.
select first_name,last_name,city,department,salary from employees where department='Marketing' and city='Delhi' and salary > 55000 
```

IN, BETWEEN, LIKE 

```sql
--Get employees in 'HR', 'Finance', or 'Marketing' 
select first_name,last_name,department from employees where department in ('HR','Finance','Marketing')
```

```sql
--Get employees with salary between 65000 and 90000
select first_name,last_name,salary from employees where salary between 65000 and 90000
```

```sql
--Get employees hired between '2018-01-01' and '2020-12-31'
select first_name,last_name,hire_date from employees where hire_date between '2018-01-01' and '2020-12-31'
```

```sql
--Get employees whose first_name starts with 'S'.
select first_name from employees where first_name like 'S%'
```

```sql
--Get employees whose last_name ends with 'n'.
select last_name from employees where last_name like '%n'
```

```sql
--Get employees whose email contains 'singh' anywhere.
select email from employees where email like '%singh%'
```

```sql
--Get employees whose city is not in ('Delhi', 'Mumbai')
select first_name,last_name,city from employees where city not in('Delhi','Mumbai')
```

```sql
--Get employees where city is NULL
select first_name,last_name,city from employees where city is Null
```

WHERE — final round

```sql
--Get employees whose first_name is exactly 4 letters long and starts with 'J'
select first_name from employees where first_name like 'J___'
```

```sql
--Get employees whose email does not contain 'kumar'
select first_name,last_name,email from employees where email not like '%kumar%'
```

```sql
--Get employees in 'Engineering' with salary > 90000, OR anyone in 'Sales' regardless of salary.
select first_name,last_name,salary,department from employees where (department='Engineering' and salary >90000) or department='Sales'; 
```

```sql
----Get employees in 'Engineering' with salary > 90000, OR anyone in 'Sales' regardless of salary. (without using parenthesis)
select first_name,last_name,department,salary from employees where department='Engineering' and salary > 90000 or department='Sales'
```

```sql
--Get employees who are in ('Engineering', 'Finance'), AND have a salary BETWEEN 70000 AND 100000, AND whose email is NOT LIKE '%test%'
select first_name,last_name,department,salary,email from employees where department in('Engineering','Finance') and salary between 70000 and 100000 and email not like '%test%'
```

```sql
--Get employees whose city is IN ('Bangalore', 'Chennai') AND (status = 'Active' OR salary > 100000)
select first_name,last_name,status,city,salary from employees where city in ('Bangalore','Chennai') and (status='Active' or salary > 100000)
```
